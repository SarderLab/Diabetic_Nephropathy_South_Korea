```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    accuracy_score,
    roc_curve
)
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler


def load_data(csv_file, label_column):
    df = pd.read_csv(csv_file)
    patient_ids = df.iloc[:, 0]  # Assuming first column is patient ID
    features = df.iloc[:, 1:316]  # Adjust column range if needed

    labels = df[[patient_ids.name, label_column]].drop_duplicates(subset=patient_ids.name)
    labels = labels.set_index(patient_ids.name)

    unique_ids = patient_ids.unique()
    X = []
    y = []

    for pid in unique_ids:
        patient_data = features[patient_ids == pid].values
        if patient_data.size == 0:
            continue
        pooled = patient_data.mean(axis=0)
        X.append(pooled)
        y.append(labels.loc[pid][label_column])

    return np.array(X), np.array(y), unique_ids


def compute_specificity(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    return tn / (tn + fp) if (tn + fp) else 0.0


def plot_and_save_roc(y_true, y_probs, output_path):
    fpr, tpr, _ = roc_curve(y_true, y_probs)
    auc = roc_auc_score(y_true, y_probs)

    plt.figure()
    plt.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve")
    plt.legend()
    plt.savefig(output_path)
    plt.close()


def plot_and_save_confusion_matrix(y_true, y_pred, output_path, title="Confusion Matrix"):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False)
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")
    plt.title(title)
    plt.tight_layout()
    plt.savefig(output_path)
    plt.close()
    return cm
```

```python

def run_svm_pipeline(csv_file, label_column, output_dir="svm_output"):
    os.makedirs(output_dir, exist_ok=True)

    X, y, patient_ids = load_data(csv_file, label_column)
    scaler = StandardScaler()
    X = scaler.fit_transform(X)

    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=72)

    all_true = []
    all_pred = []
    all_probs = []

    for fold, (train_idx, test_idx) in enumerate(skf.split(X, y)):
        print(f"\n--- Fold {fold + 1} ---")
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]

        clf = SVC(kernel='rbf', probability=True, class_weight='balanced', random_state=42)
        clf.fit(X_train, y_train)

        y_pred = clf.predict(X_test)
        y_proba = clf.predict_proba(X_test)[:, 1]

        all_true.extend(y_test)
        all_pred.extend(y_pred)
        all_probs.extend(y_proba)

        # Fold metrics
        acc = accuracy_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred, zero_division=0)
        recall = recall_score(y_test, y_pred, zero_division=0)
        specificity = compute_specificity(y_test, y_pred)
        f1 = f1_score(y_test, y_pred, zero_division=0)
        mcc = matthews_corrcoef(y_test, y_pred)
        auc = roc_auc_score(y_test, y_proba)

        print(f"Accuracy:    {acc:.4f}")
        print(f"Precision:   {precision:.4f}")
        print(f"Recall:      {recall:.4f}")
        print(f"Specificity: {specificity:.4f}")
        print(f"F1 Score:    {f1:.4f}")
        print(f"MCC:         {mcc:.4f}")
        print(f"AUC:         {auc:.4f}")

    # Final evaluation
    acc = accuracy_score(all_true, all_pred)
    precision = precision_score(all_true, all_pred, zero_division=0)
    recall = recall_score(all_true, all_pred, zero_division=0)
    specificity = compute_specificity(all_true, all_pred)
    f1 = f1_score(all_true, all_pred, zero_division=0)
    mcc = matthews_corrcoef(all_true, all_pred)
    auc = roc_auc_score(all_true, all_probs)
    cm = plot_and_save_confusion_matrix(
        all_true, all_pred,
        os.path.join(output_dir, "confusion_matrix_overall.png"),
        title="Overall Confusion Matrix"
    )
    report = classification_report(all_true, all_pred, zero_division=0)

    print("\n=== Final Evaluation ===")
    print(report)
    print("Confusion Matrix:")
    print(cm)
    print(f"Accuracy:    {acc:.4f}")
    print(f"Precision:   {precision:.4f}")
    print(f"Recall:      {recall:.4f}")
    print(f"Specificity: {specificity:.4f}")
    print(f"F1 Score:    {f1:.4f}")
    print(f"MCC:         {mcc:.4f}")
    print(f"AUC:         {auc:.4f}")

    # Save reports
    with open(os.path.join(output_dir, "classification_report.txt"), "w") as f:
        f.write(report)

    with open(os.path.join(output_dir, "metrics_summary.txt"), "w") as f:
        f.write(f"Accuracy:    {acc:.4f}\n")
        f.write(f"Precision:   {precision:.4f}\n")
        f.write(f"Recall:      {recall:.4f}\n")
        f.write(f"Specificity: {specificity:.4f}\n")
        f.write(f"F1 Score:    {f1:.4f}\n")
        f.write(f"MCC:         {mcc:.4f}\n")
        f.write(f"AUC:         {auc:.4f}\n")

    plot_and_save_roc(all_true, all_probs, os.path.join(output_dir, "roc_curve.png"))
```

```python

csv_path = "/blue/pinaki.sarder/narra.sr/5_DN_SKorea/data/combined_glom_features_with_outcomes.csv"  # Replace with actual path
label_column = "esrd"                # Replace with your label column
run_svm_pipeline(csv_path, label_column, output_dir="output_v1_seed_72")
```
