```python
# rnn_pipeline.py (TF 2.x compatible with Keras)

import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve, matthews_corrcoef, precision_score, recall_score, auc
import tensorflow as tf
import csv

# --- GPU Support ---
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"[INFO] {len(gpus)} GPU(s) available. Memory growth enabled.")
    except RuntimeError as e:
        print(f"[ERROR] {e}")
else:
    print("[INFO] No GPU found, running on CPU.")

# --- Data Loading Functions ---
def import_data(csv_file):
    df = pd.read_csv(csv_file)
    patient_ids = df.iloc[:, 0]
    features = df.iloc[:, 1:208]  # Adjust columns as needed
    feats = []
    unique_ids = patient_ids.unique()
    for pid in unique_ids:
        patient_data = features[patient_ids == pid].values
        feats.append(patient_data.tolist())
    return feats

def import_labels(csv_file, label_column_name):
    df = pd.read_csv(csv_file)
    patient_ids = df.iloc[:, 0]
    labels_df = df[[patient_ids.name, label_column_name]].drop_duplicates(subset=patient_ids.name)
    labels_df = labels_df.set_index(patient_ids.name)
    unique_ids = patient_ids.unique()
    labels = labels_df.loc[unique_ids][label_column_name].values
    return labels.astype(np.int32)

# --- Feature Standardization ---
def standardize_features(feats):
    mean_record = []
    std_record = []
    for case in feats:
        mean_record.append(np.mean(np.asarray(case), axis=0))
        std_record.append(np.std(np.asarray(case), axis=0))
    feature_mean = np.nanmean(mean_record, axis=0)
    feature_std = np.nanstd(std_record, axis=0)
    feature_std = np.nan_to_num(feature_std)
    feature_mean[feature_mean == np.inf] = 0

    standardized_feats = []
    for patient in feats:
        patient = np.array(patient)
        patient -= feature_mean
        patient /= feature_std
        patient = np.nan_to_num(patient)
        standardized_feats.append(patient)
    return standardized_feats

# --- Balanced Random Batcher ---
def balanced_random_batcher(feats, time_step, batch_size, labels, seed=None):
    rng = random.Random(seed) if seed is not None else random
    num_features = len(feats[0][0])

    # Indices per class
    class0_idx = [i for i, lbl in enumerate(labels) if lbl == 0]
    class1_idx = [i for i, lbl in enumerate(labels) if lbl == 1]

    half_batch = batch_size // 2
    batched_data = np.zeros((batch_size, time_step, num_features))
    batched_labels = np.zeros((batch_size, 2))

    # Sample half batch from class 0
    for i in range(half_batch):
        idx = rng.choice(class0_idx)
        case = feats[idx]
        label = labels[idx]
        batched_labels[i, label] = 1
        for j in range(time_step):
            batched_data[i, j, :] = rng.choice(case)

    # Sample half batch from class 1
    for i in range(half_batch, batch_size):
        idx = rng.choice(class1_idx)
        case = feats[idx]
        label = labels[idx]
        batched_labels[i, label] = 1
        for j in range(time_step):
            batched_data[i, j, :] = rng.choice(case)

    # If batch_size is odd, add one more sample from majority class or randomly
    if batch_size % 2 == 1:
        idx = rng.choice(class0_idx + class1_idx)
        case = feats[idx]
        label = labels[idx]
        batched_labels[-1, label] = 1
        for j in range(time_step):
            batched_data[-1, j, :] = rng.choice(case)

    return batched_data, batched_labels

def generate_full_sequence_batch(feats, time_step):
    batch_x = []
    for patient in feats:
        if len(patient) >= time_step:
            indices = np.linspace(0, len(patient) - 1, time_step, dtype=int)
            seq = np.array(patient)[indices]
        else:
            # Pad if sequence is too short
            seq = np.pad(patient, ((0, time_step - len(patient)), (0, 0)), mode='edge')
        batch_x.append(seq)
    return np.array(batch_x)

# --- RNN Model (Keras) ---
def build_model(num_input, seq_len, num_classes, drop_rate):
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(seq_len, num_input)),
        tf.keras.layers.Dense(num_input, activation=tf.nn.leaky_relu),
        tf.keras.layers.LSTM(75, return_sequences=True),
        tf.keras.layers.Dropout(drop_rate),
        tf.keras.layers.LSTM(25),
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python
def run_kfold_training(csv_file, label_column, seq_len=50, drop=0.3, num_classes=2, training_steps=200,
                       batch_size=32, save_interval=50, learning_rate=0.001, output_root="output_seq_50_v2"):

    raw_features = import_data(csv_file)
    all_features = np.array(standardize_features(raw_features), dtype=object)
    all_labels = np.array(import_labels(csv_file, label_column))

    skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

    all_true = []
    all_pred = []
    all_probas = []

    for fold, (train_idx, test_idx) in enumerate(skf.split(all_features, all_labels)):
        print(f"\n[INFO] Starting Fold {fold + 1}/10")

        train_feats = all_features[train_idx]
        train_lbls = all_labels[train_idx]
        test_feats = all_features[test_idx]
        test_lbls = all_labels[test_idx]

        num_input = len(train_feats[0][0])
        model = build_model(num_input, seq_len, num_classes, drop)

        fold_dir = os.path.join(output_root, f"fold_{fold+1}")
        os.makedirs(fold_dir, exist_ok=True)

        metrics_file = os.path.join(fold_dir, "metrics.csv")
        with open(metrics_file, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(["step", "accuracy", "precision", "recall", "specificity", "auc", "mcc", "loss"])

        loss_history = []

        for step in range(training_steps + 1):
            batch_x, batch_y = balanced_random_batcher(train_feats, seq_len, batch_size, train_lbls)
            loss, acc = model.train_on_batch(batch_x, batch_y)
            loss_history.append(loss)

            if step % save_interval == 0 or step == training_steps:
                val_batch_size = (len(test_feats) // 2) * 2 or len(test_feats)
                val_x, val_y = balanced_random_batcher(test_feats, seq_len, val_batch_size, test_lbls)
                preds_val = model.predict(val_x)
                y_true = np.argmax(val_y, axis=1)
                y_pred = np.argmax(preds_val, axis=1)
                probas = preds_val[:, 1]

                acc = np.mean(y_true == y_pred)
                prec = precision_score(y_true, y_pred, zero_division=0)
                rec = recall_score(y_true, y_pred, zero_division=0)
                spec = recall_score(y_true, y_pred, pos_label=0, zero_division=0)
                try:
                    auc = roc_auc_score(y_true, probas)
                except ValueError:
                    auc = float('nan')
                mcc = matthews_corrcoef(y_true, y_pred)

                print(f"[Fold {fold+1} | Step {step}] Acc: {acc:.4f}, Prec: {prec:.4f}, Rec: {rec:.4f}, Spec: {spec:.4f}, AUC: {auc:.4f}, MCC: {mcc:.4f}, Loss: {loss:.4f}")

                with open(metrics_file, "a", newline="") as f:
                    writer = csv.writer(f)
                    writer.writerow([step, acc, prec, rec, spec, auc, mcc, loss])

                plt.figure(figsize=(6, 5))
                plt.plot(range(0, step + 1), loss_history, label="Training Loss", color="orange")
                plt.xlabel("Training Step")
                plt.ylabel("Loss")
                plt.gca().set_facecolor("white")
                plt.title(f"Fold {fold+1} - Loss up to Step {step}")
                plt.legend()
                loss_plot_path = os.path.join(fold_dir, f"loss_up_to_step_{step}.png")
                plt.savefig(loss_plot_path)
                plt.close()

        # Final evaluation for fold
        full_val_x = generate_full_sequence_batch(test_feats, seq_len)
        full_val_y = tf.keras.utils.to_categorical(test_lbls, num_classes=2)
        full_preds = model.predict(full_val_x)
        full_y_true = np.argmax(full_val_y, axis=1)
        full_y_pred = np.argmax(full_preds, axis=1)
        full_probas = full_preds[:, 1]

        all_true.extend(full_y_true)
        all_pred.extend(full_y_pred)
        all_probas.extend(full_probas)

        print(f"\n[Fold {fold+1} Evaluation]")
        print(classification_report(full_y_true, full_y_pred, zero_division=0))
        print("Confusion Matrix:")
        print(confusion_matrix(full_y_true, full_y_pred))

    # Save predictions
    predictions_csv = os.path.join(output_root, "all_folds_predictions.csv")
    with open(predictions_csv, "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["true_label", "predicted_label", "predicted_probability"])
        for t, p, prob in zip(all_true, all_pred, all_probas):
            writer.writerow([t, p, prob])
    print(f"[INFO] Saved all fold predictions to {predictions_csv}")

    # ROC Curve
    fpr, tpr, _ = roc_curve(all_true, all_probas)
    roc_auc = roc_auc_score(all_true, all_probas)
    plt.figure(figsize=(6, 6))
    plt.plot(fpr, tpr, color="blue", label=f"ROC Curve (AUC = {roc_auc:.4f})")
    plt.plot([0, 1], [0, 1], color="gray", linestyle="--", label="Chance")
    plt.xlabel("False Positive Rate", color="black")
    plt.ylabel("True Positive Rate", color="black")
    plt.title("ROC Curve Across All Folds", color="black")
    plt.tick_params(colors="black")
    plt.legend(loc="upper left", facecolor="white", edgecolor="black", labelcolor="black")
    plt.gca().set_facecolor("white")  # White plot area
    roc_path = os.path.join(output_root, "final_roc_curve.png")
    plt.savefig(roc_path, facecolor="white")  # Save with white background
    plt.close()

    # Confusion Matrix
    conf = confusion_matrix(all_true, all_pred)
    plt.figure(figsize=(5, 4))
    plt.imshow(conf, interpolation="nearest", cmap=plt.cm.Blues)
    plt.title("Confusion Matrix", color="black")
    plt.colorbar()
    tick_marks = np.arange(num_classes)
    class_labels = [f"Class {i}" for i in tick_marks]
    plt.xticks(tick_marks, class_labels, color="black")
    plt.yticks(tick_marks, class_labels, color="black")
    plt.xlabel("Predicted Label", color="black")
    plt.ylabel("True Label", color="black")
    for i in range(num_classes):
        for j in range(num_classes):
            if i==0 and j==0:
                plt.text(j, i, conf[i, j], ha="center", va="center", color="white")
            else:
                plt.text(j, i, conf[i, j], ha="center", va="center", color="black")
                
    plt.gca().set_facecolor("white")  # White plot area
    conf_path = os.path.join(output_root, "final_confusion_matrix.png")
    plt.tight_layout()
    plt.savefig(conf_path, facecolor="white")  # Save with white background
    plt.close()

    # Classification Report + Overall Metrics
    report = classification_report(all_true, all_pred, digits=4, zero_division=0)
    overall_acc = np.mean(np.array(all_true) == np.array(all_pred))
    overall_auc = roc_auc_score(all_true, all_probas)
    overall_mcc = matthews_corrcoef(all_true, all_pred)
    overall_spec = recall_score(all_true, all_pred, pos_label=0, zero_division=0)

    report_path = os.path.join(output_root, "final_classification_report.txt")
    with open(report_path, "w") as f:
        f.write("Classification Report:\n")
        f.write(report)
        f.write("\n\nOverall Metrics Across All Folds:\n")
        f.write(f"Overall Accuracy: {overall_acc:.4f}\n")
        f.write(f"Overall AUC: {overall_auc:.4f}\n")
        f.write(f"Overall MCC: {overall_mcc:.4f}\n")
        f.write(f"Overall Specificity: {overall_spec:.4f}\n")

    print(f"[INFO] Saved final classification report and metrics to {report_path}")
```

```python
csv_path = "/blue/pinaki.sarder/narra.sr/5_DN_SKorea/data/combined_tubular_features_with_outcomes.csv"
label = "3y_composite_outcome"
run_kfold_training(
        csv_file=csv_path,
        label_column=label,
        seq_len=30,
        drop=0.4,
        num_classes=2,
        training_steps=200,
        batch_size=8,
        save_interval=50,
        learning_rate=0.002,
        output_root="output_seq_30_drop_0.4_lr_0.002_v1_seed_42"
    )
```
